```python
import pandas as pd
import seaborn as sns
import os
import matplotlib.pyplot as plt

sns.set_theme()
```

```python
os.chdir('/Users/felixdiederichs/PycharmProjects/Data_Analysis/.venv/Analyse/Obesity')

data = pd.read_csv('obesity_fairness.csv')

data.rename(columns = {'Epochs ': 'Epochs', 'Lambda ': 'Lambda'}, inplace = True)

data_tabfairgan = data[data['Model'] == 'tabfairgan']
data_distcorrgan = data[data['Model'] == 'distcorrgan']
data_multifairgan = data[data['Model'] == 'multifairgan']
data_decaf = data[data['Model'] == 'decaf']
data_tvae = data[data['Model'] == 'tvae']
data_ctgan = data[data['Model'] == 'ctgan']
```

```python
sns.boxplot(x = 'Model', y = 'Statistical Parity Difference', data = data)
```

```python
data_tabfairgan = data_tabfairgan.sort_values(by=['Epochs', 'Lambda'], ascending=False)
 
sns.lineplot(x='Lambda', y='Statistical Parity Difference', hue = 'Epochs', data=data_tabfairgan, errorbar=None)
```

```python
data_distcorrgan = data_distcorrgan.sort_values(by=['Epochs', 'Lambda'], ascending=False)

sns.lineplot(x='Lambda', y='Statistical Parity Difference', hue= 'Epochs', data = data_distcorrgan, errorbar=None)
```

```python
data_multifairgan = data_multifairgan.sort_values(by=['Epochs', 'Lambda'], ascending=False)

sns.lineplot(x='Lambda', y='Statistical Parity Difference', hue= 'Epochs', data = data_multifairgan, errorbar=None)
```

```python
data_tvae = data_tvae.sort_values(by=['Epochs'], ascending=False)

sns.lineplot(x='Epochs', y='Statistical Parity Difference', data=data_tvae, errorbar=None)
```

```python
data_ctgan = data_ctgan.sort_values(by='Epochs', ascending=False)

sns.lineplot(x='Epochs', y='Statistical Parity Difference', data = data_ctgan, errorbar=None)
```

```python
data_sort = data.sort_values(by=['Epochs'], ascending=False)

sns.lineplot(x='Epochs', y='Statistical Parity Difference', hue='Model', data = data_sort, errorbar=None)
```
