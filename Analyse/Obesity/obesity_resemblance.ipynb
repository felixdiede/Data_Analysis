```python
import pandas as pd
import seaborn as sns
import os
import matplotlib.pyplot as plt

pd.set_option('display.width', None)
pd.set_option('display.max_rows', None)

sns.set_theme()
```

```python
os.chdir("/Users/felixdiederichs/PycharmProjects/Data_Analysis/.venv/Analyse/Obesity/")

data = pd.read_csv("obesity_resemblance.csv")

print(data.head())

data_tabfairgan = data[data['Model'] == 'tabfairgan']
data_distcorrgan = data[data['Model'] == 'distcorrgan']
data_multifairgan = data[data['Model'] == 'multifairgan']
data_decaf = data[data['Model'] == 'decaf']
data_tvae = data[data['Model'] == 'tvae']
data_ctgan = data[data['Model'] == 'ctgan']
```

```python
sns.boxplot(x = 'Model', y = 'Proportion positive numerical tests', data = data)
```

```python
sns.lineplot(x = 'Epochs', y = 'Proportion positive numerical tests', data = data_tabfairgan, errorbar=None)
sns.lineplot(x = 'Epochs', y = 'Proportion positive numerical tests', data = data_distcorrgan, errorbar=None)
sns.lineplot(x= 'Epochs', y = 'Proportion positive numerical tests', data = data_multifairgan, errorbar=None)
sns.lineplot(x= 'Epochs', y = 'Proportion positive numerical tests', data = data_tvae, errorbar=None)
sns.lineplot(x= 'Epochs', y = 'Proportion positive numerical tests', data = data_ctgan, errorbar=None)
sns.lineplot(x= 'Epochs', y = 'Proportion positive numerical tests', data = data_decaf, errorbar=None)
plt.legend(labels=['TabFairGAN', 'DistCorrGAN', 'MultiFairGAN', 'DECAF', 'TVAE', 'CTGAN'], fontsize=7)
```
